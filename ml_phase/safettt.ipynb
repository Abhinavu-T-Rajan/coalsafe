```python
import pandas as pd
from sklearn.preprocessing import StandardScaler,LabelEncoder
import joblib
from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score

from sklearn.preprocessing import LabelEncoder,OneHotEncoder,OrdinalEncoder
```

```python
df = pd.read_csv('minesafe_100_rows_10_features.csv')
df
```

```python
df.isnull().sum()
```

```python
df['incident_type'].unique()
```

```python
df['incident_type'] = df['incident_type'].replace({'Water Ingress': 'Water Increase'})
df.to_csv("minesafe_100_rows_10_features.csv", index=False)
```

```python
df
```

```python
df['severity'].unique()
```

```python
df['communication_status'].unique()
```

```python
le = LabelEncoder()
df['incident_type']=le.fit_transform(df['incident_type'])
df
```

```python
ord = OrdinalEncoder(categories=[['Normal','Degraded','Lost']])
df['communication_status']=ord.fit_transform(df[['communication_status']])
df
```

```python
ord = OrdinalEncoder(categories=[['Minor', 'Moderate', 'Serious', 'Critical']])
df['severity']=ord.fit_transform(df[['severity']])
df
```

```python
df['priority'].unique()
```

```python
ord = OrdinalEncoder(categories=[['LOW','MEDIUM', 'HIGH', 'CRITICAL']])
df['priority']=ord.fit_transform(df[['priority']])
df
```

```python
x= df.drop(['priority'],axis=1)
x
```

```python
y = df[['priority']]
y
```

```python
Scaler=StandardScaler()
df1=Scaler.fit_transform(x)
df1
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,random_state=42,test_size=0.2)
```

```python
x_train_Scaled=Scaler.fit_transform(x_train)
x_test_Scaled=Scaler.fit_transform(x_test)
```

```python
x_train
```

```python
x_test
```

```python
from sklearn.svm import SVC
```

```python
from sklearn.metrics import accuracy_score
```

```python
svm_model=SVC(kernel='linear')
```

```python
svm_model.fit(x_train_Scaled,y_train)
```

```python
svm_prediction=svm_model.predict(x_test_Scaled)
```

```python
accuracy_score(y_test,svm_prediction)
```

```python
joblib.dump(svm_model,"minesafe_priority_model.pkl")
```
